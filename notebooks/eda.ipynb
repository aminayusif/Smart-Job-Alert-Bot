## Imports & Setup

```python
# Standard libraries
import pandas as pd
import numpy as np
import re
import string

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 200)
```

## Load Dataset

```python
# Load raw dataset
df = pd.read_csv("../data/raw/job_postings.csv")

print("Dataset shape:", df.shape)
df.head()
```

## Select relevant columns

```python
# Example column names
df = df[["Job Title", "Job Description"]]

df.head()
```

## Drop Missing & Duplicates

```python
# Remove missing
df.dropna(subset=["Job Title", "Job Description"], inplace=True)

# Remove duplicates
df.drop_duplicates(inplace=True)

print("After cleaning shape:", df.shape)
```

## Text Cleaning Function

```python
def clean_text(text):
    # Lowercase
    text = text.lower()
    
    # Remove URLs
    text = re.sub(r"http\S+", "", text)
    
    # Remove numbers
    text = re.sub(r"\d+", "", text)
    
    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))
    
    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()
    
    return text
```

```python
df["clean_title"] = df["Job Title"].apply(clean_text)
df["clean_description"] = df["Job Description"].apply(clean_text)

df.head()
```

## Combine Text Fields

```python
df["text"] = df["clean_title"] + " " + df["clean_description"]
```

## Basic Text Analysis

### Average description length

```python
df["text_length"] = df["text"].apply(lambda x: len(x.split()))

df["text_length"].describe()
```

## Top 20 most common words

```python
from collections import Counter

all_words = " ".join(df["text"]).split()
word_counts = Counter(all_words)

word_counts.most_common(20)
```

## Final Dataset Selection

We now keep only necessary columns for ML.

```python
df.rename(columns={
    "Job Title": "title",
    "Job Description": "description"
}, inplace=True)

final_df = df[["title", "description", "text"]].copy()

print(final_df.shape)
final_df.head()
```

## Save Processed Dataset

```python
final_df.to_csv("../data/processed/job_postings_cleaned.csv", index=False)

print("Processed dataset saved successfully.")
```
